### Task 6
Generate embeddings for two different sentences and compare their similarity.

```python
import numpy as np
def cosine_similarity(vec1, vec2):
    return np.dot(vec1, vec2) / (np.linalg.norm(vec1) * np.linalg.norm(vec2))
sentence1 = "Artificial Intelligence is transforming the world."
sentence2 = "AI is changing industries across the globe."
response1 = client.models.embed_content(
    model="gemini-embedding-001",
    contents=sentence1
)
embedding1 = np.array(response1.embeddings[0].values)
response2 = client.models.embed_content(
    model="gemini-embedding-001",
    contents=sentence2
)
embedding2 = np.array(response2.embeddings[0].values)
similarity = cosine_similarity(embedding1, embedding2)

print("Sentence 1:", sentence1)
print("Sentence 2:", sentence2)
print("Cosine Similarity:", similarity)
```

### Task 7
Compute cosine similarity between embeddings of three different sentences and identify the closest pair.

```python
import numpy as np
def cosine_similarity(vec1, vec2):
    return np.dot(vec1, vec2) / (np.linalg.norm(vec1) * np.linalg.norm(vec2))
sentences = [
    "I love machine learning.",
    "I enjoy studying artificial intelligence.",
    "The weather is very hot today."
]
embeddings = []
for sentence in sentences:
    response = client.models.embed_content(
        model="gemini-embedding-001",
        contents=sentence
    )
    emb = np.array(response.embeddings[0].values)
    embeddings.append(emb)
similarities = {}
for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):
        sim = cosine_similarity(embeddings[i], embeddings[j])
        similarities[(i, j)] = sim
closest_pair = max(similarities, key=similarities.get)
print("Pairwise Similarities:")
for pair, score in similarities.items():
    print(f"{sentences[pair[0]]}  <-->  {sentences[pair[1]]}  =  {score}")
print("\nClosest Pair:")
print(sentences[closest_pair[0]])
print(sentences[closest_pair[1]])
print("Similarity Score:", similarities[closest_pair])
```

### Task 10
Train and compare Logistic Regression, KNN, and Decision Tree models on a classification dataset and evaluate accuracy.

```python
pip install scikit-learn numpy
```

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
data = load_iris()
X = data.data
y = data.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)
log_model = LogisticRegression(max_iter=200)
knn_model = KNeighborsClassifier(n_neighbors=3)
dt_model = DecisionTreeClassifier()
log_model.fit(X_train, y_train)
knn_model.fit(X_train, y_train)
dt_model.fit(X_train, y_train)
log_pred = log_model.predict(X_test)
knn_pred = knn_model.predict(X_test)
dt_pred = dt_model.predict(X_test)
print("Logistic Regression Accuracy:",
      accuracy_score(y_test, log_pred))
print("KNN Accuracy:",
      accuracy_score(y_test, knn_pred))
print("Decision Tree Accuracy:",
      accuracy_score(y_test, dt_pred))
```
